### Query 2: Make new email addresses

This makes an email address from each employee’s name. It changes spaces to dots, uses lowercase letters, and adds `@ndongowater.gov`. It only shows the new emails; it does not save them yet.

```python
%%sql
SELECT
CONCAT(LOWER(REPLACE(employee_name,' ','.')), '@ndongowater.gov') AS new_email
FROM 
employee
```

### Query 3: Save the email addresses

This updates the `email` column for **every employee** using the new email format. **Note:** `UPDATE` changes data in the database.

```python
%%sql
UPDATE employee
SET email= CONCAT(LOWER(REPLACE(employee_name,' ','.')), '@ndongowater.gov') 
```

### Query 4: Check the new emails

This shows the email addresses after the update, so we can check if they were saved.

```python
%%sql
SELECT 
email
from
employee
```

# Number of Records Per Town

```python
%%sql
SELECT
Distinct town_name,
COUNT(town_name) AS number_of_records

FROM location
GROUP BY town_name
ORDER BY number_of_records DESC;
```

```python
%%sql
SELECT 
    type_of_water_source,
    COUNT(type_of_water_source) AS number_of_sources
FROM water_source
GROUP BY type_of_water_source
```

```python
%%sql
SELECT 
    type_of_water_source,
    COUNT(type_of_water_source) AS number_of_sources,
    SUM(number_of_people_served) AS number_of_sharing_people
FROM water_source
GROUP BY type_of_water_source
ORDER BY  number_of_sharing_people DESC;
```

```python
%%sql
SELECT 
    type_of_water_source,    
    COUNT(type_of_water_source) AS number_of_sources, 
    AVG(number_of_people_served) AS number_of_sharing_people,
    -- Calculates percentage of total population (27,628,140)
    ROUND(AVG(number_of_people_served) * 100.0 / 27628140, 0) AS Percentage_of_sharing
FROM 
    water_source
GROUP BY 
    type_of_water_source
ORDER BY 
    number_of_sharing_people DESC;
```

```python
%%sql
SELECT 
    type_of_water_source,
    SUM(number_of_people_served) AS number_of_sharing_people,
    RANK() OVER (ORDER BY SUM(number_of_people_served) DESC) AS rank_by_population
FROM 
    water_source
GROUP BY type_of_water_source
ORDER BY number_of_sharing_people DESC;
```

```python
%%sql
SELECT 
    DAYNAME(time_of_record) AS day_of_week,
    ROUND(AVG(time_in_queue), 0) AS avg_queue_time
FROM 
    visits
WHERE 
    time_in_queue > 0
GROUP BY 
    day_of_week;

```

```python
%%sql
SELECT 
    TIME_FORMAT(TIME(time_of_record), '%H:00') AS hour_of_day,
    ROUND(AVG(time_in_queue), 0) AS avg_queue_time
FROM 
    visits
WHERE 
    time_in_queue > 0
GROUP BY 
    hour_of_day;
```

```python
%%sql
SELECT
TIME_FORMAT(TIME(time_of_record), '%H:00') AS hour_of_day,
-- Sunday
ROUND(AVG(
CASE
WHEN DAYNAME(time_of_record) = 'Sunday' THEN time_in_queue
ELSE NULL
END
),0) AS Sunday,
-- Monday
ROUND(AVG(
CASE
WHEN DAYNAME(time_of_record) = 'Monday' THEN time_in_queue
ELSE NULL
END
),0) AS Monday,
ROUND(AVG(
CASE
WHEN DAYNAME(time_of_record) = 'Tuesday' THEN time_in_queue
ELSE NULL
END
),0) AS Tuesday,
ROUND(AVG(
CASE
WHEN DAYNAME(time_of_record) = 'Wednesday' THEN time_in_queue
ELSE NULL
END
),0) AS Wednesday,
ROUND(AVG(
CASE
WHEN DAYNAME(time_of_record) = 'Saturday' THEN time_in_queue
ELSE NULL
END
),0) AS Saturday
FROM
visits
WHERE
time_in_queue != 0 -- this excludes other sources with 0 queue times
GROUP BY
hour_of_day
ORDER BY
hour_of_day;
```
